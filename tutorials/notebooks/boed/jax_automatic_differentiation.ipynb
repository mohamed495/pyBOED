## **TUTO ON AUTOMATIC DIFFERATIATION WITH JAX**

```python
import jax
import jax.numpy as jnp
import numpy as np
import matplotlib.pyplot as plt 
import sympy as sp
```

### First example : single variable 

Let's take $f(x) = x^3 + 2x^2 -3x + 1$,

By definition, $f'(x) = \lim\limits_{h \rightarrow 0} \frac{f(x+h)-f(x)}{h}$, so here $f'(x) = 3x² + 4x - 3$

```python
def f(x) : 
    return x**3  + 2*x**2 -3*x + 1

def fprime(x) : 
    return 3*x**2 + 4*x - 3
```

```python
x = np.linspace(-5, 5, 100)
y = f(x=x)
yp = fprime(x)

plt.plot(x, y)
plt.plot(x, yp)
plt.plot(x,np.zeros(100))
plt.show()
```

```python
x = np.linspace(-5, 5, 100)
y = f(x=x)
fprime_jax = jax.grad(f)
fprime_vmap = jax.vmap(fprime_jax) # vectorial
yp = fprime_vmap(x) # np.array([fprime_jax(xi) for xi in x])
plt.plot(x, y)
plt.plot(x, yp)
plt.plot(x,np.zeros(100))
plt.show()
```
